```python
import string

# turkish stop words list 
with open('stop-words_tr.txt', encoding='utf-8') as f:
    stop_words = set(line.strip().lower() for line in f if line.strip())


def has_numbers(word):
    """sayı değeri varsa true döndür"""
    return any(char.isdigit() for char in word)

def is_unit(word):
    return len(word) <= 2

def separate_words(sent):
    sent = sent.translate(str.maketrans('', '', string.punctuation))
    splitted = sent.lower().split()
    result = []
    for word in splitted:
        w = word.strip()
        if w not in stop_words and not has_numbers(w) and not is_unit(w):
            result.append(w)
    return result
```

```python
df['product_url'] = df['product_url'].apply(lambda x: x[0] if isinstance(x, list) else x)
df['product_url'] = df['product_url'].str.strip()

# -----------------------------------------------
MAX_FEATURES = 10_000
N_COMPONENTS = 100
BATCH_SIZE = 10_000

print(f"Toplam ürün sayısı: {len(df)}")

tfidf = TfidfVectorizer(max_features=MAX_FEATURES)
tfidf.fit(df['Bow'])

sample_size = min(100_000, len(df))
sample_texts = df['Bow'].sample(sample_size, random_state=42).tolist()
tfidf_sample = tfidf.transform(sample_texts)

svd = TruncatedSVD(n_components=N_COMPONENTS, random_state=42)
svd.fit(tfidf_sample)

d = N_COMPONENTS
nlist = 100
quantizer = faiss.IndexFlatIP(d)
index = faiss.IndexIVFFlat(quantizer, d, nlist, faiss.METRIC_INNER_PRODUCT)

sample_reduced = svd.transform(tfidf_sample).astype(np.float32)
faiss.normalize_L2(sample_reduced)

print("Index training...")
index.train(sample_reduced)
print("Index training tamamlandı!")

for start in range(0, len(df), BATCH_SIZE):
    end = min(start + BATCH_SIZE, len(df))
    batch_texts = df['Bow'].iloc[start:end].tolist()
    
    tfidf_batch = tfidf.transform(batch_texts)
    svd_batch = svd.transform(tfidf_batch).astype(np.float32)
    faiss.normalize_L2(svd_batch)
    
    index.add(svd_batch)
print("Tüm veri Faiss indexine eklendi!")
```

```python
def recommend(user_input, min_price=0, max_price=1e6, top_n=5):
    input_tfidf = tfidf.transform([user_input])
    input_svd = svd.transform(input_tfidf).astype(np.float32)
    faiss.normalize_L2(input_svd)

    D, I = index.search(input_svd, top_n * 3)

    results_df = df.iloc[I[0]].copy()
    results_df['score'] = D[0]
    results_df = results_df[(results_df['price'] >= min_price) & (results_df['price'] <= max_price)]

    results_df = results_df.sort_values(by='score', ascending=False).head(top_n)

    results = []
    for _, row in results_df.iterrows():
        results.append({
            "model": row['title'],
            "price": row['price'],
            "rating": row.get('rating', 'N/A'),
            "category": ' '.join(row['category']) if isinstance(row['category'], list) else row['category'],
            "supercategory": ' '.join(row['supercategory']) if isinstance(row['supercategory'], list) else row['supercategory'],
            "product_url": row['product_url'] if isinstance(row['product_url'], str) else '',
            "score": round(row['score'], 4)
        })
    
    return results
```

```python
query = "bebek puseti"
min_price = 500
max_price = 2000

recommendations = recommend(query, min_price, max_price, top_n=5)
for r in recommendations:
    print(r)
```

```python
def recommend_with_model(user_input, model_name, top_n=5):
    model = models[model_name]
    index = indices[model_name]

    input_embedding = model.encode([user_input], convert_to_numpy=True, normalize_embeddings=True)
    D, I = index.search(input_embedding, top_n)

    results = []
    for idx, score in zip(I[0], D[0]):
        row = df.iloc[idx]
        results.append({
            "model": row['title'],
            "price": row['price'],
            "score": round(score, 4),
            "url": row.get('product_url', ''),
            "rate" : row['weighted_rating_label_2']
        })
    return results
```

```python
from sentence_transformers import SentenceTransformer
import time
import os
import pickle

def encode_and_index_in_batches(model, texts, batch_size=10000):
    first_batch = True
    index = None

    for i in range(0, len(texts), batch_size):
        batch_texts = texts[i:i+batch_size]
        batch_embeddings = model.encode(
            batch_texts,
            convert_to_numpy=True,
            normalize_embeddings=True
        )

        if first_batch:
            dim = batch_embeddings.shape[1]
            index = faiss.IndexFlatIP(dim)
            first_batch = False

        index.add(batch_embeddings)

    return index

model_names = [
    'paraphrase-multilingual-MiniLM-L12-v2',
    'paraphrase-MiniLM-L6-v2',
]

texts = df['Bow'].tolist()
models = {}
indices = {}

model_times = {}

save_dir = 'saved_models'
os.makedirs(save_dir, exist_ok=True)

for name in model_names:
    print(f"\n Model yükleniyor: {name}")
    start_time = time.time()

    model_file = os.path.join(save_dir, f"{name.replace('/', '_')}.pt")
    index_file = os.path.join(save_dir, f"{name.replace('/', '_')}.index")

    if os.path.exists(model_file) and os.path.exists(index_file):
        print("Kayıtlı model ve index bulunuyor, yükleniyor...")
        model = SentenceTransformer(model_file, device='cuda')
        index = faiss.read_index(index_file)
    else:
        model = SentenceTransformer(name, device='cuda')
        print("Embedding ve FAISS indexleme başlatıldı...")
        index = encode_and_index_in_batches(model, texts, batch_size=10000)
        model.save(model_file)
        faiss.write_index(index, index_file)
        print("Model ve index dosyaya kaydedildi.")

    elapsed_time = time.time() - start_time
    model_times[name] = elapsed_time

    models[name] = model
    indices[name] = index

    print(f"{name} tamamlandı. Süre: {round(elapsed_time / 60, 2)} dakika")

print("\nMODELLERİN İŞLEM SÜRELERİ:")
for model_name, t in model_times.items():
    print(f"{model_name}: {round(t / 60, 2)} dakika")
```

```python
def clean_turkish_chars(val):
    turkish_map = str.maketrans("çğıöşüÇĞİÖŞÜ", "cgiosuCGIOSU")

    if isinstance(val, list):
        return [v.translate(turkish_map) if isinstance(v, str) else v for v in val]
    elif isinstance(val, str):
        return val.translate(turkish_map)
    return val

df_finally = df.applymap(clean_turkish_chars)
df_finally.to_csv("all_data_cleaned.csv", index=False)
df_finally.head()
```
